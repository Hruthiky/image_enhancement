```python
  from google.colab import files
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
import numpy as np

# =========================================================
# 🔹 1. ALWAYS FORCE USER TO UPLOAD A NEW IMAGE
# =========================================================
print("📤 Please upload an input image...")
uploaded = files.upload()
input_image_path = list(uploaded.keys())[0]
print("👉 Uploaded:", input_image_path)

# =========================================================
# 🔹 2. IMAGE ENHANCEMENT FUNCTION
# =========================================================
def enhance_image(img):
    # Convert to LAB for lighting enhancement
    lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
    l, a, b = cv2.split(lab)

    # CLAHE improves brightness & contrast for degraded images
    clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8,8))
    cl = clahe.apply(l)

    enhanced_lab = cv2.merge((cl, a, b))
    enhanced = cv2.cvtColor(enhanced_lab, cv2.COLOR_LAB2BGR)

    # Sharpening to restore edges
    kernel = np.array([[0, -1, 0],
                       [-1, 5, -1],
                       [0, -1, 0]])
    enhanced = cv2.filter2D(enhanced, -1, kernel)

    return enhanced

# =========================================================
# 🔹 3. LOAD & ENHANCE IMAGE
# =========================================================
img = cv2.imread(input_image_path)
enhanced_img = enhance_image(img)

enhanced_path = "enhanced_input.jpg"
cv2.imwrite(enhanced_path, enhanced_img)

# =========================================================
# 🔹 4. LOAD BOTH MODELS (Custom + Pretrained)
# =========================================================
custom_model = YOLO("/content/runs/detect/train/weights/best.pt")   # Corrected path to trained model
pretrained_model = YOLO("yolov8n.pt")                              # Official YOLOv8n

# =========================================================
# 🔹 5. RUN DETECTIONS ON ENHANCED IMAGE
# =========================================================
custom_results = custom_model.predict(enhanced_path, save=True)
pretrained_results = pretrained_model.predict(enhanced_path, save=True)

# Read saved output paths
custom_output_path = custom_results[0].save_dir + "/" + custom_results[0].path.split("/")[-1]
pretrained_output_path = pretrained_results[0].save_dir + "/" + pretrained_results[0].path.split("/")[-1]

# =========================================================
# 🔹 6. SHOW: Original → Enhanced → Custom → Pretrained
# =========================================================
plt.figure(figsize=(20, 14))

# Original image
plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Input Image")
plt.axis("off")

# Enhanced image
plt.subplot(2, 2, 2)
plt.imshow(cv2.cvtColor(enhanced_img, cv2.COLOR_BGR2RGB))
plt.title("Enhanced Image")
plt.axis("off")

# Custom model detection
plt.subplot(2, 2, 3)
custom_img = cv2.imread(custom_output_path)
plt.imshow(cv2.cvtColor(custom_img, cv2.COLOR_BGR2RGB))
plt.title("Detection: Custom Model")
plt.axis("off")

# Pretrained YOLO detection
plt.subplot(2, 2, 4)
pre_img = cv2.imread(pretrained_output_path)
plt.imshow(cv2.cvtColor(pre_img, cv2.COLOR_BGR2RGB))
plt.title("Detection: Pretrained YOLOv8n")
plt.axis("off")

plt.show()

print("✔ Enhanced image saved as:", enhanced_path)
print("✔ Custom model results saved in:", custom_results[0].save_dir)
print("✔ Pretrained model results saved in:", pretrained_results[0].save_dir)
```
